### Extracting the evacuation times

Let's create a helper function to extract the relevant information from the raw records!

```python
# helper function
def extractEvacTime(s, totalSteps):
    bits = s.split(',')
    record = {}
    for b in bits:
        try:
            k,v = b.split(':')
            if k[0] not in record:
                record[k] = float(v)
        except:
            pass

    if "FINISHED_PREP_START_EVACUATING" in record:
        startTime = record["FINISHED_PREP_START_EVACUATING"]
    elif "BEGIN_EVACUATING" in record:
        startTime = record["BEGIN_EVACUATING"]
    else:
        return -1
        
    finishTime = totalSteps

    if "FINISH_EVAC" in record:
        finishTime = record["FINISH_EVAC"]
    elif "ENTER_SHELTER" in record:
        finishTime = record["ENTER_SHELTER"]
    elif "DONE" in record:
        finishTime = record["DONE"]
    elif "TRAPPED" in record:
        finishTime = record["TRAPPED"]
        
    return finishTime - startTime

def getAllValidEvacs(df, total_sim_time):
    raw = [extractEvacTime(s, total_sim_time) for s in df["evacuatingRecord"]]
    a = np.array(raw)
    return a[a >= 0]
```

### Reading in the files

Ok, go through the files and pull out the evacuating times.

```python
evacueeFullData = []
evacuatingFullData = []
trappedFullData = []
numTrapped = []
vertical = []
total_evacuating = []

scenario_names = []
prep_times = []

# extract information from the raw files
for f in files:
    
    # there are some files in the same directory that we don't want to process here - skip those
    
    if (not (f.startswith("out_params_default") and f.endswith("txt")) # only read default param scenarios
        or "trackedAgents" in f  # ignore files tracking agents
        or "roaduse" in f        #        "        "     road usage
        or "heatmap" in f):      #        "        "     spatial presence
        continue
    #print(f) # which one are we processing?
    
    # otherwise, create a scenario name from the file's qualities
    
    # first, does the scenario include vertical evacuation?
    s_name = "vert_" if "_vertEvac" in f else "hori_"

    # secondly, how much prep time is involved? Extract from the filename
    prep_time = 30
    if "min" in f:
        min_index = f.find("min")
        prep_time = int(f[min_index - 2:min_index]) # must convert to an integer!
    s_name += str(prep_time)

    # save this information to the record, with quick lookups for instances of particular scenarios
    scenario_names.append(s_name)
    prep_times.append(prep_time)
    vertical.append("_vertEvac" in f)

    # read in the file
    
    raw_data = pd.DataFrame(pd.read_csv(workingDir + f, sep='\t')) 

    # extract the data into a series of arrays for easier reference
    
    total_evacuating.append(raw_data[raw_data["evacuatingRecord"].notna()].shape[0])
    
    # pull out the successful evacuees
    evacuees = raw_data[raw_data["status"] == "Evacuated"].reset_index()
    evacueeFullData.append(getAllValidEvacs(evacuees, total_sim_time))
    
    #   "    " those who haven't completed their evacuation at the end, but also haven't been inundated
    evacuees = raw_data[raw_data["status"] == "Evacuating"].reset_index()
    evacuatingFullData.append(getAllValidEvacs(evacuees, total_sim_time))
    
    #   "    " those who HAVE been inundated
    trapped = raw_data[raw_data["status"] == "Trapped"].reset_index()
    trappedFullData.append(trapped)
    
    # there are MANY trapped individuals, so for easier handling, a quick look-up for how many people are trapped
    numTrapped.append(len(trapped))
       
#len(evacueeFullData)
if(len(evacueeFullData) == 0):
    print("Did you remember to update the filename prefix?")
else:
    print(f"Range of individuals caught in the tsunami: \n{numTrapped}")
```

### By Attribute

Make it possible to extract by attributes

```python
def selectOnAttribute(df, targetValues, waterFile = None, exclude=False):
    
    myDf = df
    
    # select on the flooding scenario?
    if waterFile is not None:
        # pull out the runs on a specific water setting
        myDf = df[df["waterShapefile"] == waterFile]

    # make sure there's actually something to select!
    if len(targetValues) == 0:
        return myDf
        
    # iterate through, selecting on these
    for i, j in targetValues.items():
        
        if exclude:
            myDf = myDf[~myDf[i].isin(j)]
            
        else:
            myDf = myDf[myDf[i].isin(j)]            

    return myDf
```
